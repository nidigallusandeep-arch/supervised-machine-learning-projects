# Handle missing values

```python
df["Age"]=df["Age"].fillna(df["Age"].mean())
df["Experience"]=df["Experience"].fillna(df["Experience"].mean())
```

```python
df.duplicated().sum()
```

```python
df.isnull().sum()
```

# Select numerical features

```python
X=df[["Age","Experience","Salary"]]
print(X)
```

# MinMaxScaler

```python
from sklearn.preprocessing import MinMaxScaler
min_max=MinMaxScaler()
X_min_max=min_max.fit_transform(X)
print(X_min_max)
```

```python
minmax_df = pd.DataFrame(X_min_max,columns=["Age", "Experience", "Salary"])
minmax_df
```

# RobustScaler

```python
from sklearn.preprocessing import RobustScaler
robust=RobustScaler()
X_robust=robust.fit_transform(X)
X_robust
```

```python
robust_df = pd.DataFrame(X_robust,columns=["Age", "Experience", "Salary"])
robust_df
```
